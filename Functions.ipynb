1. What is the difference between a function and a method in Python?
-  Function:
A function is a block of code that performs a specific task and is defined independently. It is not tied to any particular object or class. Functions are called directly by their name and can accept arguments as input and optionally return a value.

- Method:
A method is also a block of code that performs a specific task, but it is associated with an object (an instance of a class). Methods are defined within a class and are called on an object using dot notation (e.g., object.method()). The first argument of a method is conventionally named self, which refers to the instance of the class the method is called on, allowing it to access and modify the object's attributes.

2. Explain the concept of function arguments and parameters in Python?

Arguments are the actual values that are passed to the function when it is invoked.
-  In the add function, num1 and num2 are parameters that specify the function expects two inputs. When the function is called with add(5, 3), the values 5 and 3 are arguments that provide the actual data for the parameters.
- Python supports different types of arguments:
Positional arguments:
These are passed to the function in the order they are defined in the function signature.
Keyword arguments:
These are passed to the function with the parameter name specified, allowing for arguments to be passed in any order.
Default arguments:
These are parameters that have a default value, which is used if no argument is passed for that parameter in the function call.
Variable-length arguments:
These allow a function to accept an arbitrary number of positional or keyword arguments using the *args and **kwargs syntax.

```python
# parameters
def greet(name):
    print(f"Hello, {name}!")
#arguments
    greet("Alice")
```

5. What are iterators in Python and how do they differ from iterables?
-  In Python, an iterable is any object that can be looped over, like lists, tuples, strings, and dictionaries. An iterator is an object that provides a way to sequentially access the elements of an iterable.
-  Iterables:
-Definition: An object that implements the __iter__ method, which returns an iterator.
-Example: Lists, tuples, strings, sets, dictionaries, and custom classes can all be iterables.
-Purpose: To be looped over using a for loop or accessed with the next() function and the iter() method.
-  Iterators:
-Definition: An object that implements the __iter__ and __next__ methods.
-Example: The iterator returned by calling iter() on an iterable, or a custom iterator class.
-Purpose: To provide the next element in a sequence when called with next(). It maintains internal state, remembering where it is in the sequence.
-State: Iterators have internal state, so you can only iterate through them once.
-StopIteration: When an iterator has no more elements, calling next() will raise a StopIteration exception.

```python
# Iterables
my_list = [1, 2, 3, 4, 5]
for item in my_list:
    print(item)

# Iterators
my_iterator = iter(my_list)
print(next(my_iterator))
print(next(my_iterator))

```

6. Explain the concept of generators in Python and how they are defined.
-  In Python, generators are a type of iterator that allows you to produce a sequence of values on-the-fly, without storing the entire sequence in memory at once. They are defined using functions that include the yield keyword instead of return. When a generator function is called, it returns a generator object, which is an iterator. This object can then be iterated over to produce the sequence of values, one at a time.
- A generator function is defined using the def keyword, like a regular function.
- Instead of return, it uses yield, which pauses the function’s execution and retains its state.
- When resumed, execution continues from where it left off.




```python
# Generators
def countdown(n):
    while n > 0:
        yield n
        n -= 1

counter = countdown(5)
for num in counter:
    print(num)
```

7. What are the advantages of using generators over regular functions?
-  Generators offer several advantages over regular functions when dealing with large datasets, streamlining code execution and memory usage. Here’s why they shine:
1. Memory Efficiency
- Regular functions store all values in memory before returning them.
- Generators produce values on demand, avoiding high memory consumption.
- Example: Instead of storing a large list, a generator creates items as needed.

2. Lazy Evaluation
- Generators pause execution after each yield statement and resume when requested.
- This is useful for handling large sequences, such as reading big files or streaming data.

3. Better Performance
- Since generators don’t precompute values, they reduce CPU overhead.
- For loops and other iteration-based operations run more efficiently.

4. Easy Implementation of Iterators
- Traditional iterators require defining __iter__() and __next__().
- Generators simplify iteration without extra methods.
5. Infinite Sequences:
Generators can be used to create infinite sequences of values, unlike regular functions that need to return a complete result.






8. What is a lambda function in Python and when is it typically used?
-  A lambda function is a small, anonymous function defined using the lambda keyword.
It can have any number of arguments but can only have a single expression.
Lambda functions are often used for short, simple operations where defining a full function using def would be overkill.
-  Lambda functions are typically used in scenarios where:

-  - You need a small, one-time use function: Instead of defining a separate function using def, you can use a lambda function for a quick operation.
-  - You want to pass a function as an argument to another function: Higher-order functions like map, filter, and sorted often take functions as arguments. Lambda functions provide a concise way to define these functions inline.
-  - You want to create closures or decorators: Lambda functions can be used to create closures, which are functions that remember the values in enclosing scopes. They are also used in creating decorators, which are functions that modify the behavior of other functions.

```python
square = lambda x: x * x
print(square(5))
```

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
-  map(), filter(), and reduce() are built-in higher-order functions in Python that operate on iterables (like lists, tuples, etc.) and are often used in functional programming. They each serve distinct purposes:
-- map():
Applies a given function to each item in an iterable and returns a new iterator (map object) with the results. It transforms each element individually.
-- filter():
Creates a new iterator containing only the elements from the original iterable for which a given function returns True. It selects elements based on a condition.
-- reduce():
Applies a function cumulatively to the items of an iterable, from left to right, to reduce the iterable to a single value. It performs an aggregate computation.


```python
numbers = [1,2,3,4,5,6]
#map()
def square(x):
    return x * x

squared_numbers = map(square, numbers)
print(list(squared_numbers))

#reduce()
from functools import reduce

def add(x, y):
    return x + y

sum_result = reduce(add, numbers)
print(sum_result)

#filter()
def is_even(x):
    return x % 2 == 0

even_numbers = filter(is_even, numbers)
print(list(even_numbers))
```

```python
#  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
class FibonacciIterator:
    def __init__(self, limit):
        self.limit = limit
        self.a, self.b = 0, 1
        self.count = 0
        def __iter__(self):
          return self
        def __next__(self):
          if self.count >= self.limit:
            raise StopIteration
        result = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return result
        limit = 10
        fibonacci_iterator = FibonacciIterator(limit)
        for num in fibonacci_iterator:
            print(num)
            print(limit)
```

```python
# Create a Python program that uses `filter()` to remove all the vowels from a given string.
def remove_vowels(input_string):
    vowels = 'aeiouAEIOU'
    filtered_string = ''.join(filter(lambda char: char not in vowels, input_string))
    return filtered_string
    input_string = "Hello, World!"
    result = remove_vowels(input_string)
    print(f"The string without vowels is: {result}")
    print(input_string)
```
